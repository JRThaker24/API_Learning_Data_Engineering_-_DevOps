## 1.1 What Is an API? — The Core Idea

**API** stands for **Application Programming Interface**.

Break the three words down in plain English:

- **Application** → a piece of software (a website, a mobile app, a database, a weather service).
- **Programming** → meant to be used *by code*, not by a human clicking buttons.
- **Interface** → a well-defined "plug" or "contract" that says *"send me a request in THIS shape, and I'll send you a response in THAT shape."*

> **One-line definition:** *An API is a contract that lets one piece of software talk to another piece of software in a predictable way.*

Think of it as a **messenger with rules**. You (the caller) don't need to know *how* the other software works inside. You only need to know:
1. **Where** to send the message (the address / URL).
2. **What** to send (the request format).
3. **What** you'll get back (the response format).

That "you don't need to know how it works inside" part has a name: **abstraction**. The API *hides complexity* and exposes only what you need.

```
   YOU (your program)                  SOMEONE ELSE'S SYSTEM
   ┌───────────────┐                   ┌───────────────────────┐
   │  "Give me the │   ── request ──▶  │  Huge complex system   │
   │   weather in  │                   │  (databases, sensors,  │
   │   Mumbai"     │   ◀── response ─  │   models, servers...)  │
   └───────────────┘                   └───────────────────────┘
        You only see the plug ↑             All hidden behind the API ↑
```

**Why "interface"?** A power socket is an interface. You plug in any device; you don't rewire your house. The socket hides the power plant, the grid, the transformers. An API is the software version of that socket.


## 1.2 Why APIs Exist & Why They Matter

### The problem APIs solve
Before APIs were common, if App A wanted data from App B, developers either:
- copied App B's entire codebase and database (a nightmare to keep in sync), or
- got a raw dump of a database (insecure, fragile, and it breaks the moment B changes a table).

APIs solve this by offering a **stable, controlled door**: B decides exactly what A can see and do, and can change everything *behind* the door without breaking A — as long as the door (the contract) stays the same.

### Why they matter (the business + engineering view)
| Reason | What it means |
|--------|----------------|
| **Reuse** | Build once, let many apps consume it (web, mobile, partners). |
| **Separation of concerns** | Frontend team and backend team work independently against a contract. |
| **Security** | Expose *only* what's needed; hide the database and internal logic. |
| **Integration** | Stripe (payments), Twilio (SMS), OpenAI (AI), AWS (cloud) — you plug in instead of rebuilding. |
| **Scalability** | Different parts scale independently (microservices). |
| **Monetization** | Companies *sell* API access (this is a huge business model). |

### 🏭 Data Engineering lens (read this carefully — it recurs all course)
A Data Engineer's job is often described as **"move data from where it is to where it's useful."** APIs are one of the three most common *sources* and *sinks* of data (the others being files and databases). Concretely:
- You **pull** data from SaaS tools (Salesforce, Stripe, Shopify, Google Analytics, Twitter/X) **through their APIs**.
- You **expose** curated datasets or ML features to other teams **as an API**.
- You **trigger** and **monitor** pipelines (Airflow, Databricks) **via APIs**.

If you cannot confidently work with APIs, you cannot be a Data Engineer. That's why this course exists.


## 1.3 A Short History of APIs (so the jargon makes sense)

You don't need dates memorized, but the *evolution* explains why so many "styles" exist.

```
1960s-70s   Subroutine libraries         "call this function in the same program"
   │
1980s-90s   RPC (Remote Procedure Call)  "call a function on ANOTHER machine"
   │        + CORBA / DCOM (complex, language-specific)
   │
2000        SOAP + WSDL + XML            "enterprise web services", heavy, strict
   │
2000        Roy Fielding's PhD thesis    defines REST (an architectural STYLE)
   │
2005-2015   REST + JSON explosion        simple, web-friendly, mobile-friendly → dominant
   │
2015        GraphQL (Facebook)           "ask for exactly the fields you want"
   │
2016        gRPC (Google)                fast binary RPC over HTTP/2, great for microservices
   │
Today       REST is still the default; GraphQL & gRPC used where they fit.
```

**Takeaway:** REST won the public-API world because it is simple and rides on HTTP (the web's own protocol). SOAP still lives in banks/enterprises. gRPC lives *inside* companies between microservices. GraphQL is popular for flexible frontends. As a Data Engineer you'll meet **mostly REST**, occasionally GraphQL (e.g., GitHub, Shopify) and gRPC (internal systems).


## 1.4 Five Real-World Analogies

Analogies are how you'll *explain APIs in interviews*. Master at least two.

### 🍽️ 1. The Restaurant Analogy (the classic)
```
   YOU (customer)      WAITER (API)          KITCHEN (server/database)
   ┌──────────┐        ┌──────────┐          ┌────────────────────┐
   │  "I want │  ───▶  │  takes    │  ───▶   │  chefs cook food    │
   │  a pizza"│        │  order    │         │  (you never see it) │
   │          │  ◀───  │  brings   │  ◀───   │  plates the dish    │
   │  eats 🍕 │        │  food     │         │                     │
   └──────────┘        └──────────┘          └────────────────────┘
```
- **You** = the client app.
- **Menu** = the API documentation (what you're allowed to order).
- **Waiter** = the API (takes your request, returns a response).
- **Kitchen** = the server + database (hidden complexity).
- You don't enter the kitchen. You don't need to know the recipe. You just order from the menu. **That's an API.**

### 🏦 2. The Bank / ATM Analogy
An ATM is an API to the bank. You insert a card + PIN (**authentication**), request ₹5000 (**request**), and get cash (**response**). You never touch the bank's core ledger (**database**). The ATM enforces rules: correct PIN, sufficient balance, daily limit (**rate limiting**). If you enter the wrong PIN 3 times, it blocks you (**error + security**).

### 📦 3. The Courier / Delivery Analogy
You give a courier a **parcel with an address label** (the **request**: body + headers + destination). The courier network figures out routing (**you don't care how**). You get a **tracking status** (200 delivered, 404 address not found, 500 truck crashed). **Idempotency**: sending the *same* parcel twice by accident should not charge you twice — good systems detect duplicates.

### ✈️ 4. The Airport Analogy
- **Boarding pass** = **token** (proves you're allowed on this flight).
- **Security check** = **authentication** (who are you?).
- **Gate access for business lounge** = **authorization** (what are you allowed to do?).
- **Flight schedule board** = **API documentation**.
- **Air traffic control** = **load balancer / gateway** (routes planes so runways don't collide).

### 🔌 5. The Power Socket Analogy (best for "abstraction")
Any device with the right plug works in the socket. You don't rewire the grid. The socket is a **standard interface**. Change the power plant (coal → solar) and your laptop still charges. That's an API decoupling the consumer from the implementation.


## 1.5 How Software Actually Communicates

Two programs talk by **sending messages over a network** using an agreed **protocol** (a set of rules). For web APIs that protocol is almost always **HTTP** (covered fully in Module 2).

The universal pattern is **request → response**:

```
   CALLER (client)                                CALLEE (server)
      │                                                 │
      │   1. open a connection to server address        │
      │────────────────────────────────────────────────▶
      │                                                 │
      │   2. send a REQUEST                             │
      │      (method + address + headers + body)        │
      │────────────────────────────────────────────────▶
      │                                                 │  3. server does work
      │                                                 │     (query DB, compute)
      │   4. send a RESPONSE                            │
      │      (status code + headers + body)             │
      │◀────────────────────────────────────────────────
      │                                                 │
      │   5. connection closes (or is reused)           │
```

Key mental model: **the caller always starts** (in REST). The server *responds*; it does not spontaneously call you back. (The exception — where the server calls *you* — is a **Webhook**, covered in 1.9.)

**Synchronous vs asynchronous:**
- *Synchronous*: you send a request and **wait** for the response (like a phone call).
- *Asynchronous*: you send a request and **continue working**; the result arrives later (like a text message / webhook / message queue). Data pipelines use both.


## 1.6 Client vs Server

These two words appear thousands of times in your career. Nail them now.

| | **Client** | **Server** |
|---|-----------|-----------|
| Role | *Asks* for something | *Provides* something |
| Starts the conversation? | ✅ Yes | ❌ No (it waits) |
| Examples | Browser, mobile app, your Python script, `curl`, Postman | Flask app, database server, weather service |
| In the restaurant | Customer | Kitchen |

```
   ┌────────────┐        request         ┌────────────┐
   │   CLIENT   │  ───────────────────▶  │   SERVER   │
   │ (asks)     │  ◀───────────────────  │ (answers)  │
   └────────────┘        response        └────────────┘
```

**Important nuance:** *client* and *server* are **roles, not machines.** The same program can be a server to some and a client to others.

> 🏭 **DE example:** Your Airflow task is a **client** when it calls the Stripe API to pull payments. That same pipeline later becomes a **server** when it exposes the cleaned data through an internal "metrics API" that the analytics team calls. Roles flip depending on direction.


## 1.7 Frontend vs Backend (and where the API sits)

```
   ┌─────────────────────┐        HTTP API         ┌──────────────────────┐
   │      FRONTEND       │  ───────────────────▶   │       BACKEND        │
   │  (what the user     │     JSON request        │  (logic + database)  │
   │   sees & clicks)    │  ◀───────────────────   │                      │
   │  HTML/CSS/JS,       │     JSON response       │  Flask/Django/Node,  │
   │  React, mobile app  │                         │  PostgreSQL, Redis   │
   └─────────────────────┘                         └──────────────────────┘
                              ▲
                              │
                     THE API is this boundary — the contract
                     between frontend and backend.
```

- **Frontend** = presentation. Runs in the browser or on the phone. Cannot be trusted (users can tamper with it) → **never** put secrets or final validation only in the frontend.
- **Backend** = the brain. Validates, applies business rules, talks to databases, enforces security.
- **The API is the negotiated contract between them.**

> 🏭 **DE reality:** Data Engineers live almost entirely in **backend + data layer**. You rarely build UIs. But you constantly build the **APIs the frontend/analysts consume**, and you consume other backends' APIs.


## 1.8 Monolith vs Microservices vs API Gateway

### Monolith
One big application; all features live in one codebase and deploy together.
```
   ┌───────────────────────────────────────┐
   │              MONOLITH                  │
   │  [Users] [Orders] [Payments] [Emails]  │
   │        one codebase, one deploy        │
   └───────────────────────────────────────┘
```
✅ Simple to start, easy to debug locally. ❌ Hard to scale one piece; one bug can crash everything; big teams step on each other.

### Microservices
Split into small independent services, each owning one job, each with its **own API** (and often its own database). They talk to each other over the network.
```
   ┌────────┐   ┌────────┐   ┌──────────┐   ┌────────┐
   │ Users  │   │ Orders │   │ Payments │   │ Emails │
   │  API   │   │  API   │   │   API    │   │  API   │
   └───┬────┘   └───┬────┘   └────┬─────┘   └───┬────┘
       └──── talk over HTTP/gRPC to each other ──┘
```
✅ Scale/deploy independently, teams work in parallel. ❌ Network complexity, harder to trace, distributed failures.

### API Gateway
When you have many microservices, you don't want clients calling 20 different addresses. An **API Gateway** is a single front door that routes requests to the right service and handles cross-cutting concerns.
```
                         ┌──────────────────┐
   Client ───request──▶  │   API GATEWAY    │  ── auth, rate-limit, routing, logging
                         └───────┬──────────┘
                 ┌───────────────┼───────────────┐
              ┌──▼───┐        ┌───▼───┐       ┌───▼────┐
              │Users │        │Orders │       │Payments│
              └──────┘        └───────┘       └────────┘
```
The gateway typically handles: **authentication, rate limiting, routing, request/response transformation, logging, caching**. Examples: AWS API Gateway, Kong, NGINX, Apigee.

> 🏭 **DE relevance:** Data platforms are increasingly microservice/gateway-based. You may pull from an **ingestion service API**, push to a **feature-store API**, and everything flows through a gateway that enforces auth + rate limits. Understanding this shape helps you debug *where* a pipeline is failing.


## 1.9 API Styles: REST, SOAP, GraphQL, gRPC, RPC, Webhooks

These are **different ways** to design APIs. REST is the star of this course; know the others for interviews.

### RPC (Remote Procedure Call) — the oldest idea
"Call a function that lives on another machine as if it were local."
```
   client code:  result = add(2, 3)     # but add() runs on a remote server
```
Feels like calling a function. The style focuses on **actions/verbs** (`getUser`, `createOrder`).

### SOAP (Simple Object Access Protocol)
Strict, XML-only, enterprise protocol. Has a formal contract file (**WSDL**), built-in standards for security (**WS-Security**) and transactions. Verbose but rigorous.
```xml
<soap:Envelope>
  <soap:Body>
    <getEmployee><id>42</id></getEmployee>
  </soap:Body>
</soap:Envelope>
```
✅ Strong contracts, great for banking/telecom/legacy. ❌ Heavy, verbose, not web/mobile friendly.

### REST (Representational State Transfer) — the modern default ⭐
An **architectural style** (not a protocol) that uses plain HTTP. You act on **resources** (nouns) using HTTP methods (verbs). Full details in Module 3.
```
   GET    /employees/42      → read employee 42
   POST   /employees         → create an employee
   DELETE /employees/42      → delete employee 42
```
✅ Simple, uses HTTP, JSON-friendly, cacheable, universally supported. ❌ Can over-fetch/under-fetch data; many round-trips for related data.

### GraphQL — "ask for exactly what you want"
A single endpoint; the client sends a **query** describing precisely which fields it needs.
```graphql
query {
  employee(id: 42) { name, department { name } }
}
```
✅ No over/under-fetching, one round-trip for nested data. ❌ Caching is harder, server complexity, easy to write expensive queries. (GitHub & Shopify expose GraphQL APIs you may consume as a DE.)

### gRPC — fast binary RPC over HTTP/2
Uses **Protocol Buffers** (compact binary) and a strict `.proto` schema. Great **between microservices** where speed matters.
✅ Very fast, small payloads, streaming, strong typing. ❌ Not human-readable, not directly browser-friendly.

### Webhooks — "don't call us, we'll call you" (reverse direction!)
Normal APIs: *you poll* ("any new data yet? ... yet? ... yet?"). A **webhook** flips it: you register a URL, and the provider **calls your URL** when an event happens.
```
   Normal (polling):   You ──"anything new?"──▶ Server   (repeat every minute 😩)
   Webhook (push):     Event happens → Server ──POST──▶ Your URL   (instant 🎉)
```
> 🏭 **DE relevance:** Stripe fires a webhook to your endpoint the instant a payment succeeds — far more efficient than polling their API every minute. You'll build webhook receivers to trigger pipelines in real time. (Polling vs webhooks is revisited deeply in Module 17.)

### Quick comparison
| Style | Transport | Format | Best for |
|-------|-----------|--------|----------|
| SOAP | HTTP/other | XML | Enterprise/legacy, strict contracts |
| REST | HTTP | usually JSON | Public/web/mobile APIs (default) |
| GraphQL | HTTP | JSON | Flexible frontends, nested data |
| gRPC | HTTP/2 | Protobuf (binary) | Internal microservices, speed |
| Webhook | HTTP | usually JSON | Event notifications (push) |


## 1.10 Library vs API vs SDK (people confuse these constantly)

| Term | What it is | Analogy |
|------|-----------|---------|
| **Library** | Code you import and run **inside your own program** (same machine, no network). e.g. `pandas`. | A cookbook on your shelf — you cook. |
| **API** | A **contract to talk to another system**, often over a network. | A phone number you call to order food. |
| **SDK** (Software Development Kit) | A **bundle** = one or more libraries + tools + docs + examples that make using an API easy in a specific language. e.g. `boto3` is the AWS SDK for Python. | A full meal-kit: ingredients + recipe + tools. |

Relationship:
```
   ┌──────────────────────────────────────────────┐
   │  SDK (e.g. boto3, stripe-python)               │
   │   ├── libraries (helper code)                  │
   │   ├── auth handling                            │
   │   └── nice functions that call ...             │
   │                                                │
   │        ▼ under the hood makes HTTP calls to    │
   │   ┌──────────────────────────────┐             │
   │   │   the actual REST API         │            │
   │   └──────────────────────────────┘             │
   └──────────────────────────────────────────────┘
```

> 🏭 **DE example:** `boto3.client("s3").list_buckets()` looks like a library call, but under the hood the **SDK** builds an HTTP **request** to the AWS S3 **API** and parses the **response** for you. Knowing this helps you debug: when boto3 fails, the real error is an HTTP status code from the API.


## 1.11 OpenAPI, Swagger & Postman (documentation + tooling)

You cannot use an API you don't understand. These tools describe and exercise APIs.

- **OpenAPI Specification (OAS)** — a standard, machine-readable format (YAML/JSON) that *describes* a REST API: its endpoints, methods, parameters, request/response shapes, and auth. It's the **blueprint**.
- **Swagger** — the popular set of tools built around OpenAPI. **Swagger UI** turns an OpenAPI file into a beautiful interactive web page where you can read *and try* the API in your browser. (You'll auto-generate this in Modules 19 & 22.)
- **Postman** — a desktop/web app to **manually build, send, save, and organize** API requests. The go-to tool for testing and exploring APIs by hand (Module 15).

```
   OpenAPI file (blueprint)  ──▶  Swagger UI (interactive docs page)
                                    │
   Postman (hand-testing) ─────────┘  both let you SEND real requests
```

> 🏭 **DE relevance:** When you onboard to a new data source, step one is *read its OpenAPI/Swagger docs* to learn endpoints, auth, pagination, and rate limits — **before** writing a line of extraction code. Postman lets you poke the API and confirm the shape of the data first.


## 1.12 Data Formats: JSON, XML, YAML, CSV

APIs must agree on *how data is written*. These are the four you'll meet. (Module 4 is 100% JSON.)

### JSON — JavaScript Object Notation (the API king 👑)
Human-readable, lightweight, maps perfectly to Python `dict`/`list`.
```json
{
  "id": 42,
  "name": "Asha",
  "skills": ["python", "sql"],
  "active": true,
  "manager": null
}
```

### XML — eXtensible Markup Language (older, verbose, SOAP's format)
```xml
<employee>
  <id>42</id>
  <name>Asha</name>
</employee>
```

### YAML — YAML Ain't Markup Language (config-friendly, indentation-based)
Rarely a *response* format; heavily used for **config** (Docker Compose, Kubernetes, OpenAPI, Airflow).
```yaml
employee:
  id: 42
  name: Asha
  skills:
    - python
    - sql
```

### CSV — Comma-Separated Values (tabular, spreadsheet-friendly)
```
id,name,active
42,Asha,true
43,Ravi,false
```

### When to use what
| Format | Strengths | Typical use |
|--------|-----------|-------------|
| **JSON** | Compact, nests well, universal | 90% of REST API bodies |
| **XML** | Strict, schema support | SOAP, legacy enterprise |
| **YAML** | Human-friendly config | DevOps/config files, not API payloads |
| **CSV** | Tabular, tiny, Excel-friendly | Bulk data exports, flat datasets |

> 🏭 **DE relevance:** You'll constantly convert between these. A common pipeline: **pull JSON from an API → flatten nested JSON → write CSV/Parquet to a data lake → load into a warehouse.** Module 4 teaches JSON↔dict deeply, and Module 17 teaches flattening nested API JSON into tables.


## 1.13 Anatomy of a Request & a Response

Every HTTP API interaction has exactly two messages. Learn their parts — this is the single most important diagram in Module 1.

### The REQUEST (client → server)
```
   ┌──────────────────────────────────────────────────────────┐
   │  1. METHOD + URL       →  GET  /employees/42?fields=name   │
   │  2. HEADERS            →  Authorization: Bearer abc123      │
   │                          Content-Type: application/json     │
   │                          Accept: application/json           │
   │  3. BODY (payload)     →  { "name": "Asha" }  (POST/PUT)    │
   └──────────────────────────────────────────────────────────┘
```
- **Method** — the *verb* / intent (GET read, POST create, ...). See Module 9.
- **URL** — *where* + optional **query string** (`?fields=name`) for filters.
- **Headers** — *metadata about the message* (who am I, what format, auth token).
- **Body / Payload** — the *actual data* you're sending (only for POST/PUT/PATCH). **"Payload"** = the meaningful content of the body.

### The RESPONSE (server → client)
```
   ┌──────────────────────────────────────────────────────────┐
   │  1. STATUS CODE        →  200 OK   (or 404, 500, ...)      │
   │  2. HEADERS            →  Content-Type: application/json    │
   │                          X-RateLimit-Remaining: 59          │
   │  3. BODY (payload)     →  { "id": 42, "name": "Asha" }     │
   └──────────────────────────────────────────────────────────┘
```

### Status Codes at a glance (full table in Module 2 & 23)
| Range | Meaning | Memory hook | Example |
|-------|---------|-------------|---------|
| **1xx** | Informational | "hold on" | 100 Continue |
| **2xx** | ✅ Success | "here you go" | 200 OK, 201 Created, 204 No Content |
| **3xx** | Redirect | "look elsewhere" | 301 Moved, 304 Not Modified |
| **4xx** | ❌ **Client** error (YOUR fault) | "you messed up" | 400 Bad Request, 401 Unauthorized, 403 Forbidden, 404 Not Found, 429 Too Many Requests |
| **5xx** | 💥 **Server** error (THEIR fault) | "they messed up" | 500 Internal Server Error, 502 Bad Gateway, 503 Unavailable |

> 🧠 **Interview gold:** *"What's the difference between 401 and 403?"* → **401 Unauthorized** = "I don't know who you are (bad/missing credentials)." **403 Forbidden** = "I know who you are, but you're not allowed." *"400 vs 500?"* → 400 = the client sent something wrong; 500 = the server crashed handling a valid request.

> 🏭 **DE relevance:** Your extraction code must branch on status codes: **2xx** → parse data; **429** → back off and retry; **5xx** → retry with exponential backoff; **4xx (except 429)** → usually a bug in *your* request, don't blindly retry. This logic is Module 17's backbone.


## 1.14 Cookies, Sessions & Tokens (how servers "remember" you)

HTTP is **stateless** (next section) — the server forgets you after each request. So how does a website keep you logged in? Three mechanisms:

### Cookie
A small piece of data the **server sets** and the **browser stores** and **sends back automatically** on every future request to that site.
```
   Server → response header:  Set-Cookie: session_id=xyz789
   Browser stores it, then on next request:
   Browser → request header:  Cookie: session_id=xyz789
```

### Session
Server-side memory tied to a cookie. The cookie holds only a **session id**; the *real* data (who you are, cart contents) lives **on the server**.
```
   Cookie:  session_id=xyz789   ──┐
                                  ▼
   Server memory: { "xyz789": { user: "Asha", cart: [...] } }
```

### Token (e.g., JWT — JSON Web Token)
A **self-contained** signed string the client sends (usually in the `Authorization: Bearer <token>` header). The server can verify it **without** storing session state — great for scaling and for APIs. Deep dive in Modules 2 & 14.
```
   Request header:  Authorization: Bearer eyJhbGciOi...<signed token>
```

| | Where state lives | Good for |
|---|-------------------|----------|
| **Session (cookie)** | Server memory/DB | Traditional websites |
| **Token (JWT)** | Inside the token itself (stateless) | **APIs, mobile, microservices** |

> 🏭 **DE relevance:** Almost every data-source API authenticates with **tokens/API keys**, not cookies. You'll store them as secrets and send them as headers. Some (OAuth) tokens **expire** and must be **refreshed** — a classic pipeline failure point (Module 17).


## 1.15 Authentication vs Authorization

Two words that sound alike and are constantly confused — a guaranteed interview question.

- **Authentication (AuthN)** — *"Who are you?"* Proving identity (password, API key, token). At the airport: showing your **passport at security**.
- **Authorization (AuthZ)** — *"What are you allowed to do?"* Checking permissions. At the airport: your boarding pass says **economy**, so you can't enter the business lounge.

```
   Request ─▶ [ AuthN: who are you? ] ─▶ [ AuthZ: are you allowed? ] ─▶ do the work
                 │ fail                      │ fail
                 ▼                           ▼
              401 Unauthorized            403 Forbidden
```

> 🧠 **Memory hook:** **AuthN = N for "name/identity"**, **AuthZ = Z for "zone/permission you can enter"**. And map them to status codes: **401 = AuthN failed**, **403 = AuthZ failed**.

> 🏭 **DE relevance:** An analytics service account might **authenticate** fine but be **authorized** only to *read* certain tables. When your pipeline gets a 403, the token is valid but lacks a scope/permission — a config issue, not a credential issue.


## 1.16 Stateful vs Stateless & Idempotency

### Stateless (a core REST principle)
**Each request contains everything the server needs to handle it.** The server does **not** remember previous requests. Every request carries its own auth token and data.
```
   Stateless:  Request 1 ─ has token, has all context
               Request 2 ─ has token, has all context   (server remembers nothing between them)
```
✅ Easy to scale (any server instance can handle any request → put a load balancer in front). ❌ Slightly larger requests (must resend context/token).

### Stateful
The server **remembers** your prior interactions (like a session).
```
   Stateful:   Login → server remembers you → later requests rely on that memory
```
✅ Convenient. ❌ Harder to scale (requests must hit the *same* server that has your state, or you need shared storage).

### Idempotency (huge for reliability & Data Engineering)
An operation is **idempotent** if doing it **once or many times gives the same final result.**
```
   Idempotent:      Set balance = 100.   Do it 5 times → still 100.  ✅
   NOT idempotent:  Add 100 to balance.  Do it 5 times → +500.       ❌
```
HTTP method idempotency (memorize this table — asked constantly):
| Method | Idempotent? | Safe (no change)? |
|--------|-------------|-------------------|
| GET | ✅ Yes | ✅ Yes |
| HEAD | ✅ Yes | ✅ Yes |
| PUT | ✅ Yes | ❌ No |
| DELETE | ✅ Yes | ❌ No |
| POST | ❌ **No** | ❌ No |
| PATCH | ❌ Usually no | ❌ No |

> 🏭 **DE relevance — why you'll love idempotency:** Pipelines *retry*. If a network blip makes you re-run the same load, an **idempotent** write (e.g., "upsert row with id=42") won't create duplicates, but a naive "INSERT" will. Designing idempotent loads (upserts, dedup keys) is a **senior Data Engineering skill**. Revisited in Module 17.


## 1.17 Caching, Rate Limiting, Pagination, Versioning

Four practical concepts you'll meet on day one of consuming real APIs.

### Caching — "don't recompute what hasn't changed"
Store a response so future identical requests are served instantly (from memory/CDN) instead of hitting the server/DB again.
```
   1st request → server computes → response (also saved in cache)
   2nd request → served from cache instantly (server not bothered) ⚡
```
Controlled by headers like `Cache-Control`, `ETag`, `Last-Modified`. ✅ Faster + cheaper. ❌ Risk of **stale** (outdated) data — cache invalidation is famously hard.

### Rate Limiting — "you may call me only N times per window"
Servers cap how often you can call them to prevent abuse/overload. Exceed it → **429 Too Many Requests**, often with a `Retry-After` header.
```
   Limit: 60 requests/minute
   Request 61 within the minute → 429 Too Many Requests (Retry-After: 30)
```

### Pagination — "here's page 1 of 10,000; ask for the next"
Huge result sets are split into pages so no single response is gigantic. Two main styles:
- **Offset/limit:** `?limit=100&offset=200` (page 3). Simple but can skip/duplicate rows if data changes mid-scan.
- **Cursor:** the response returns a `next_cursor`; you pass it back to get the next page. Stable for changing data.
```
   Page 1 → returns 100 items + next_cursor="abc"
   Page 2 → GET ?cursor=abc → next 100 items + next_cursor="def"  ... until cursor is null
```

### Versioning — "don't break existing clients when you change the API"
Ship changes under a new version so old integrations keep working. Common styles: URL (`/v1/employees`, `/v2/employees`) or a header (`Accept: application/vnd.company.v2+json`).

> 🏭 **DE relevance:** These four are the *daily reality* of extraction: you **loop over paginated pages**, **respect rate limits** (sleep/backoff on 429), optionally **cache** to avoid re-pulling, and pin an **API version** so an upstream change doesn't silently break your pipeline overnight. Module 17 turns all of this into robust code.


## 1.18 Content Negotiation & Compression

### Content Negotiation — "what format do you want?"
The client says what it can accept; the server replies in a matching format.
```
   Client → header:  Accept: application/json      "please give me JSON"
   Server → header:  Content-Type: application/json  "here's JSON as requested"
```
The same endpoint could serve JSON *or* XML depending on the `Accept` header. `Content-Type` describes what a message body actually *is*; `Accept` describes what the client *wants back*.

### Compression — "send it smaller"
Large responses can be gzip/brotli-compressed to save bandwidth and time.
```
   Client → header:  Accept-Encoding: gzip
   Server → header:  Content-Encoding: gzip   (body is compressed; client auto-decompresses)
```
✅ Big speed/bandwidth win for large JSON. ❌ Tiny CPU cost to compress/decompress.

> 🏭 **DE relevance:** Pulling millions of rows? Requesting **gzip** can cut transfer time dramatically. Most HTTP libraries (`requests`) handle decompression automatically, but knowing it exists helps you optimize big extractions and read `Content-Encoding` in debugging.


## 1.19 Latency, Timeout, Retry

### Latency — "how long the round trip takes"
The delay between sending a request and getting a response, measured in milliseconds. Caused by network distance, server work, queuing. Not the same as **bandwidth** (how *much* data per second).
```
   send ──────────── network + server work ────────────▶ receive
        └──────────────── latency (e.g. 250 ms) ─────────┘
```

### Timeout — "I won't wait forever"
The maximum time a client will wait before giving up. Without a timeout, a hung server can freeze your program indefinitely.
```
   if no response within 30s → raise TimeoutError, stop waiting
```

### Retry — "try again on transient failures"
Network calls fail *sometimes* for temporary reasons. Retrying **transient** errors (timeouts, 429, 5xx) often succeeds. But retry **intelligently**:
- Use **exponential backoff**: wait 1s, then 2s, 4s, 8s... (don't hammer a struggling server).
- Add **jitter** (randomness) so many clients don't retry in sync.
- **Cap** the number of retries.
- **Only retry idempotent / safe** operations, or use an idempotency key (recall 1.16!).
```
   attempt 1 → fail → wait 1s
   attempt 2 → fail → wait 2s
   attempt 3 → fail → wait 4s
   attempt 4 → success ✅   (give up after N attempts)
```

> 🏭 **DE relevance:** *Always set a timeout* on pipeline HTTP calls (a hung extraction can block an entire DAG). Combine timeout + retry + exponential backoff to survive flaky upstreams. This exact pattern is coded in Module 17.


## 1.20 Proxy, Reverse Proxy, Load Balancer, CDN

These sit *between* client and server and shape traffic. Interviewers love the proxy vs reverse-proxy distinction.

### Forward Proxy — sits in front of the **client**
Acts on the client's behalf; the server sees the proxy, not the real client. Used for privacy, corporate filtering, caching outbound requests.
```
   Client ─▶ [ Forward Proxy ] ─▶ Internet ─▶ Server
             (hides the client)
```

### Reverse Proxy — sits in front of the **server**
Acts on the server's behalf; the client thinks it's talking to the real server. Used for TLS termination, caching, routing, security. **NGINX** is the classic example (Module 16).
```
   Client ─▶ [ Reverse Proxy ] ─▶ Server(s)
                                  (hides the servers)
```
> 🧠 **Memory hook:** *Forward proxy hides the **client**; reverse proxy hides the **server**.*

### Load Balancer — spreads traffic across many servers
A special reverse proxy that distributes requests so no single server is overwhelmed (round-robin, least-connections, etc.). Enables horizontal scaling — which is *why stateless (1.16) matters*: any server can handle any request.
```
                     ┌──▶ Server A
   Client ─▶ [ LB ] ─┼──▶ Server B
                     └──▶ Server C
```

### CDN — Content Delivery Network
A globally distributed cache that serves content from a location **near the user**, cutting latency. Great for static assets and cacheable API responses.
```
   User in India  ─▶ nearest CDN edge (Mumbai)  ─▶ (only if needed) origin server in USA
```

> 🏭 **DE relevance:** When you consume an external API, your request may traverse *their* CDN → load balancer → gateway → service. Knowing this helps you interpret errors: a **502/504** often comes from a proxy/gateway between you and the app, not the app itself. When you *serve* a data API in production, you'll put NGINX (reverse proxy) + a load balancer in front (Module 16).


## 1.21 CORS, HTTPS, TLS, SSL (security & the browser)

### HTTPS / TLS / SSL — encryption in transit
- **HTTP** = plain text; anyone on the network can read it. 🚨
- **HTTPS** = HTTP **secured** with encryption. The padlock in your browser.
- **TLS** (Transport Layer Security) = the modern protocol that does the encryption. **SSL** = TLS's older, now-deprecated predecessor — people still say "SSL" but mean **TLS**.
```
   HTTP:   "password=hunter2"  ← readable by attackers 😱
   HTTPS:  "x8#@f9...encrypted..."  ← unreadable in transit 🔒
```
How it works (simplified): a **TLS handshake** verifies the server's **certificate** (proving it's really who it claims) and agrees on encryption keys, *then* data flows encrypted. Full handshake in Module 2.

> 🧠 **Interview one-liner:** *"SSL and TLS?"* → "SSL is the old name; TLS is its modern, secure successor. We say SSL casually but use TLS."

### CORS — Cross-Origin Resource Sharing
A **browser** security rule. By default a web page from `site-a.com` **cannot** call an API on `site-b.com` via JavaScript (the "same-origin policy"). CORS is how the API **opts in** to allow specific other origins, via response headers like `Access-Control-Allow-Origin`.
```
   Page at  https://myapp.com   ──JS fetch──▶  https://api.other.com
   Browser: "different origin! is it allowed?"
   API responds:  Access-Control-Allow-Origin: https://myapp.com   ✅ allowed
   (missing/mismatched header → browser BLOCKS it → CORS error)
```
> ⚠️ **Crucial nuance for DE:** CORS is enforced **only by browsers.** Your Python script, `curl`, Postman, and Airflow **ignore CORS entirely** — so if you hit "CORS error", it's a *frontend/browser* problem, never a server-to-server pipeline problem. This distinction alone will save you hours.

   │  (Extract data FROM it)    (Push data TO it)      (Trigger/monitor)  │

### 1) API as a data **source** (Extract — the most common)

   Stripe API ──JSON──▶ [ your Python extractor ] ──▶ raw zone (S3/GCS)
                              │ paginate, retry, respect rate limits
                              ▼
                         land as JSON/Parquet
```

### 2) API as a data **sink** (Load / serve)
- Push transformed data *into* tools: reverse-ETL into Salesforce, send metrics to a dashboard service, POST features to an **ML feature store**.

### 3) API as a **control plane** (orchestration)

### A canonical API-driven ETL, end to end
```
   ┌─────────┐   paginate + retry + backoff   ┌──────────────┐
   │ SaaS    │ ──────────────────────────────▶│  EXTRACT     │  (Python + requests)
   │ REST API│   respect rate limits (429)     │  raw JSON    │
   └─────────┘                                 └──────┬───────┘
                                                      ▼
                                              ┌───────────────┐
                                              │  TRANSFORM    │  flatten nested JSON,
                                              │  validate     │  clean, dedup, cast types
                                              └──────┬────────┘
                                                     ▼
                                              ┌───────────────┐
                                              │  LOAD (upsert)│  idempotent write to
                                              │  warehouse    │  Postgres/Snowflake/BigQuery
                                              └───────────────┘
```

### The hard parts you must handle (all taught in Module 17)
| Challenge | Why it bites Data Engineers |
|-----------|------------------------------|
| **Pagination** | APIs cap page size; you must loop until done. |
| **Rate limits (429)** | Pull too fast → blocked; must throttle/backoff. |
| **Token refresh** | OAuth tokens expire mid-run → job fails at 3 AM. |

| **Idempotency** | Retries must not double-insert (upsert!). |
| **Dead Letter Queue** | Where do permanently-failing records go? |

Keep this list in mind — Module 17 converts each row into production-grade Python.

````python
# A minimal API CLIENT call — read this line by line (explained below)
import requests  # the de-facto Python HTTP client library (an SDK-like wrapper over HTTP)

URL = "https://api.github.com/repos/pallets/flask"  # the endpoint (WHERE) — Flask's own repo

# Make a GET request (the METHOD = "read"), with a timeout so we never hang forever.
response = requests.get(URL, timeout=10)  # returns a Response object (status + headers + body)

print("Status code:", response.status_code)     # 200 means success (recall 2xx table)
print("Content-Type:", response.headers.get("Content-Type"))  # a RESPONSE HEADER

data = response.json()  # DESERIALIZE the JSON body into a Python dict (Module 4 goes deep)
print("Repo full name :", data["full_name"])
print("Stars           :", data["stargazers_count"])
print("Open issues     :", data["open_issues_count"])
````

2. `URL = ...` — the **endpoint**: *where* we send the request. `api.github.com` is the host; `/repos/pallets/flask` is the **resource path** (Module 3 explains resources).
3. `requests.get(URL, timeout=10)` — send an HTTP **GET** (the "read" **method**). `timeout=10` says *"give up after 10 seconds"* (recall 1.19 — always set a timeout in pipelines). This returns a `Response` object.
4. `response.status_code` — the **status code** (recall 1.13). `200` = success.
5. `response.headers.get("Content-Type")` — a **response header**; here it tells us the body is JSON.
6. `response.json()` — **deserialize** (Module 4) the JSON text body into a Python `dict` so we can index it like `data["full_name"]`.
7. The prints pull specific fields out of the parsed dict — this is you *consuming* an API.

## ✅ Module 1 — Summary (revise this before interviews)

- An **API** is a **contract** letting software talk to software predictably; it hides complexity (**abstraction**) behind a stable door.
- APIs exist for **reuse, separation of concerns, security, integration, scalability, and monetization**.
- The universal pattern is **request → response** over a **protocol** (HTTP for web APIs). The **client asks**, the **server answers** — except **webhooks**, where the server calls you.
- **Frontend** (untrusted UI) ↔ **API contract** ↔ **Backend** (logic + DB). Data Engineers live in the backend/data layer.
- **Monolith** (one deploy) → **Microservices** (many small services, each an API) → fronted by an **API Gateway** (auth, routing, rate-limit).
- API **styles**: **REST** (default, HTTP+JSON), **SOAP** (XML enterprise), **GraphQL** (ask exact fields), **gRPC** (fast binary, internal), **RPC** (call remote functions), **Webhooks** (push).
- **Library** (runs in your process) vs **API** (talk to another system) vs **SDK** (a language bundle that calls an API for you, e.g. boto3).
- **OpenAPI** = blueprint; **Swagger** = interactive docs from it; **Postman** = hand-test requests.
- Formats: **JSON** (king), **XML** (SOAP/legacy), **YAML** (config), **CSV** (tabular).
- A **request** = method + URL + headers + body(payload); a **response** = status code + headers + body. **Status codes**: 2xx success, 4xx *your* fault, 5xx *their* fault. **401** (who are you?) vs **403** (not allowed).
- **Cookies/Sessions** (server remembers) vs **Tokens/JWT** (self-contained, stateless — used by APIs).
- **AuthN** = identity (→401); **AuthZ** = permission (→403).
- **Stateless** requests scale easily (load-balance freely). **Idempotent** operations are safe to retry (upsert, not append) — a key DE skill.
- Practical realities: **caching**, **rate limiting (429)**, **pagination (offset vs cursor)**, **versioning**, **content negotiation**, **compression**, **latency/timeout/retry with exponential backoff**.
- Traffic shapers: **forward proxy** (hides client), **reverse proxy** (hides server, e.g. NGINX), **load balancer** (spreads traffic), **CDN** (serves near the user).
- Security: **HTTPS** = encrypted HTTP via **TLS** (**SSL** = its old name). **CORS** is a **browser-only** rule — irrelevant to server-side pipelines.
- **Data Engineers use APIs as source (extract), sink (serve/load), and control plane (trigger/monitor)** — handling pagination, rate limits, token refresh, incremental loads, schema evolution, retries, idempotency, and monitoring.


## 🎤 Module 1 — Interview Questions (with model answers)

1. **What is an API in one sentence?** A contract that lets one program request services/data from another in a predictable, abstracted way.
2. **Client vs server?** Client *asks* and starts the conversation; server *answers*. They're roles, not machines — the same program can be both.
3. **REST vs SOAP?** REST is a lightweight architectural style over HTTP using (usually) JSON, acting on resources with HTTP verbs; SOAP is a strict, verbose XML *protocol* with formal contracts (WSDL), common in legacy/enterprise.
4. **REST vs GraphQL?** REST has multiple endpoints and can over/under-fetch; GraphQL has one endpoint and lets the client request exactly the fields it needs (at the cost of harder caching and query-cost control).
5. **When would you pick gRPC?** Internal, high-performance service-to-service communication where small binary payloads, HTTP/2 streaming, and strong typing matter more than human-readability.
6. **Library vs API vs SDK?** Library = code you run in-process; API = a contract to talk to another system; SDK = a language-specific bundle (libraries + tools + auth) that makes calling an API easy (e.g., boto3 calls AWS APIs).
7. **What is a webhook and how does it differ from polling?** Polling = you repeatedly ask "any new data?"; a webhook = the provider *pushes* a request to your registered URL when an event occurs — instant and efficient.
8. **401 vs 403?** 401 = not authenticated (unknown/invalid credentials); 403 = authenticated but not authorized (known, but forbidden).
9. **400 vs 500?** 400 = client sent a bad request (your fault); 500 = server crashed handling a valid request (their fault).
10. **AuthN vs AuthZ?** Authentication proves *who* you are; authorization decides *what* you may do.
11. **Stateless vs stateful, and why does REST prefer stateless?** Stateless requests carry all needed context, so any server instance can handle any request → trivial horizontal scaling behind a load balancer.
12. **What is idempotency and why does it matter for pipelines?** An idempotent op yields the same result whether run once or many times; it makes retries safe (no duplicate writes) — implement via upserts/dedup keys.
13. **How do you handle rate limiting when consuming an API?** Respect the limit, watch for 429 + `Retry-After`, throttle request rate, and use exponential backoff with jitter.
14. **Offset vs cursor pagination?** Offset (`limit/offset`) is simple but can skip/duplicate rows if data changes mid-scan; cursor pagination passes a stable pointer and is safe for changing datasets.
15. **What is CORS and does it affect a Python data pipeline?** CORS is a *browser* same-origin security mechanism controlled by response headers; it does **not** affect server-side clients like `requests`, curl, or Airflow.
16. **SSL vs TLS?** SSL is the deprecated predecessor of TLS; both encrypt data in transit — we say "SSL" colloquially but use TLS.
17. **Forward vs reverse proxy?** Forward proxy fronts the *client* (hides who's calling); reverse proxy fronts the *server* (hides the backends), e.g., NGINX doing TLS termination and load balancing.
18. **Where do Data Engineers use APIs?** As data sources (extract from SaaS), sinks (serve/push curated data), and control planes (trigger/monitor orchestration) — handling pagination, rate limits, token refresh, incremental loads, and idempotent writes.
19. **Why set a timeout on every HTTP call in a pipeline?** A hung server would otherwise block the task (and its DAG) indefinitely; a timeout converts that into a catchable, retryable error.
20. **What's the difference between `Content-Type` and `Accept` headers?** `Content-Type` describes the format of the *current* message's body; `Accept` tells the server what format the client *wants back* (content negotiation).

## ⚠️ Module 1 — Common Mistakes (and how to avoid them)

1. **Confusing "API" with "REST".** REST is *one style* of API. SOAP, GraphQL, gRPC are also APIs. Don't use the words interchangeably in interviews.
2. **Thinking client/server are physical machines.** They are **roles**. The same service is a client to one system and a server to another.
3. **Mixing up 401 and 403.** 401 = *who are you?* (bad/missing credentials). 403 = *I know you, but no.* (permission). This is a top interview trap.
4. **Assuming POST is idempotent.** It is **not**. Retrying a POST can create duplicates. Use idempotency keys or idempotent methods (PUT/upsert) when retries are possible.
5. **Forgetting timeouts.** A missing timeout is the #1 cause of "my pipeline hung all night." Always set one.
6. **Blindly retrying every error.** Retry transient errors (timeout, 429, 5xx). Do **not** retry most 4xx (e.g., 400/401/404) — the request itself is wrong; fix it.
7. **Thinking a CORS error is a server bug.** CORS is enforced by the **browser only**. In a Python/Airflow pipeline, CORS is irrelevant — if you see it, it's a frontend problem.
8. **Saying "SSL" and meaning HTTP-with-no-security.** HTTPS uses **TLS** (SSL is deprecated). Know the difference.
9. **Ignoring pagination.** Grabbing "page 1" and assuming it's the whole dataset silently drops data — a classic DE bug. Always loop until there's no next page/cursor.
10. **Confusing library vs SDK vs API.** `pandas` is a library (in-process). `boto3` is an SDK that *calls* AWS **APIs** over HTTP. Knowing this changes how you debug failures (an SDK error is often an HTTP status underneath).

## 2.1 What Is HTTP? (Plain English)

**HTTP** = **HyperText Transfer Protocol**. A **protocol** is just an agreed set of rules for how two parties exchange messages — like the etiquette of a phone call ("hello" → talk → "bye").

> **One-line definition:** *HTTP is the request/response, text-based protocol that clients and servers use to exchange messages on the web.*

Three defining properties (memorize — all three are interview favorites):

1. **Request/response** — the client sends one request; the server sends back one response. Always paired.
2. **Stateless** — (recall Module 1.16) each request is independent; HTTP itself remembers *nothing* about previous requests. Cookies/sessions/tokens are *bolted on top* to fake memory.
3. **Text-based (human-readable)** in HTTP/1.x — a request is literally lines of text you could type by hand. (HTTP/2 and /3 make it binary for speed, but the *meaning* is identical.)

```
   CLIENT                                    SERVER
     │  ── HTTP request  (plain text) ──▶      │
     │  ◀─ HTTP response (plain text) ──       │
     │        (server forgets you)             │
```

**HTTP vs HTTPS:** HTTPS is *exactly* HTTP, but the messages travel inside an encrypted **TLS** tunnel (section 2.6). Same rules, sealed envelope.

> 🏭 **DE relevance:** When your extraction script "calls an API", it is opening a TCP connection, doing a TLS handshake, and sending an HTTP request — every single time. Understanding these layers is how you diagnose whether a failure is DNS (can't resolve host), TCP (connection refused), TLS (certificate error), or HTTP (4xx/5xx). That triage skill separates juniors from seniors.


## 2.2 HTTP Versions (and why you should care)

| Version | Year | Key idea | Why it matters to you |
|---------|------|----------|------------------------|
| **HTTP/1.0** | 1996 | One request per TCP connection, then close. | Slow: new handshake every request. |
| **HTTP/1.1** | 1997 | **Persistent connections** (keep-alive), reuse one TCP connection for many requests. | Still everywhere; the version you'll read in examples. |
| **HTTP/2** | 2015 | **Multiplexing** — many requests in parallel over ONE connection; binary framing; header compression. | Faster for many small calls; transparent to your code. |
| **HTTP/3** | 2022 | Runs over **QUIC** (UDP-based), not TCP. Faster connection setup, no head-of-line blocking. | Newer; libraries adopt it gradually. |

```
   HTTP/1.1  (one connection, requests queue up):
      conn ── req1 ─ resp1 ─ req2 ─ resp2 ─ req3 ─ resp3 ──▶   (serial, head-of-line blocking)

   HTTP/2    (one connection, multiplexed):
      conn ── req1 req2 req3 ──▶
              ◀── resp1 resp3 resp2 (interleaved, parallel)
```

**Head-of-line blocking:** in HTTP/1.1, if request #1 is slow, requests behind it wait. HTTP/2 fixes this at the HTTP layer; HTTP/3 fixes it at the transport layer too (QUIC).

> 🧠 **Interview tip:** If asked "how did HTTP/2 improve performance?" → *multiplexing (parallel requests on one connection), header compression (HPACK), and server push.* You rarely choose the version manually — `requests`/browsers negotiate it — but knowing the difference explains why modern APIs feel fast.

> 🏭 **DE relevance:** When bulk-extracting from an API that supports HTTP/2, a client that multiplexes can pull many pages concurrently over one connection. For most `requests`-based pipelines you get HTTP/1.1; libraries like `httpx` add HTTP/2 support for higher throughput.


## 2.3 How a Browser Works (the big picture)

Before the layers, here's the whole journey of typing `https://api.github.com/repos/pallets/flask` and pressing Enter. We'll then zoom into each numbered step.

```
   You type a URL and hit Enter
            │
   ┌────────▼─────────┐
   │ 1. URL PARSING   │  scheme=https, host=api.github.com, path=/repos/pallets/flask
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 2. DNS LOOKUP    │  api.github.com  ──▶  140.82.121.6   (name → IP address)
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 3. TCP HANDSHAKE │  open a reliable connection to 140.82.121.6 : 443
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 4. TLS HANDSHAKE │  verify certificate, agree on encryption keys (because https)
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 5. HTTP REQUEST  │  GET /repos/pallets/flask  (+ headers)  sent through the tunnel
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 6. SERVER WORK   │  routing, auth, query DB, build response
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 7. HTTP RESPONSE │  200 OK + headers + JSON body
   └────────┬─────────┘
   ┌────────▼─────────┐
   │ 8. RENDER/USE    │  browser paints page  /  your script parses JSON
   └──────────────────┘
```

A **URL** breaks down like this (you must be able to name every part):
```
   https :// api.github.com : 443 / repos/pallets/flask ? per_page=5 # section
   └─┬─┘    └──────┬──────┘  └┬┘ └────────┬──────────┘ └────┬─────┘ └──┬──┘
  scheme       host         port        path              query      fragment
 (protocol)  (domain)   (default 443  (resource)      (parameters)  (client-only,
             + DNS)      for https)                                  not sent to server)
```

> 🧠 **Interview classic:** *"What happens when you type a URL and press Enter?"* — walk this diagram: URL parse → DNS → TCP → TLS → HTTP request → server processing → HTTP response → render. Mentioning all layers signals seniority.


## 2.4 Step 1 — DNS: Turning a Name into an Address

Computers talk over the network using **IP addresses** (e.g., `140.82.121.6`), not names. **DNS** (Domain Name System) is the internet's **phone book**: it translates a human name (`api.github.com`) into an IP address.

**Analogy:** You know your friend's *name* but need their *phone number* to call. DNS is the contacts app that looks it up.

```
   Browser: "what's the IP for api.github.com?"
        │
        ▼
   [ 1. OS cache ]  ── already known? return it (fast!)
        │ no
        ▼
   [ 2. Recursive resolver ]  (usually your ISP or 8.8.8.8 / 1.1.1.1)
        │ asks the hierarchy if not cached:
        ▼
   [ 3. Root servers ]  ── "ask the .com servers"
        ▼
   [ 4. TLD servers (.com) ]  ── "ask github.com's nameservers"
        ▼
   [ 5. Authoritative NS ]  ── "api.github.com = 140.82.121.6"
        │
        ▼
   Answer cached at each level for a while (TTL), then returned.
```

Key terms:
- **Resolver** — the server that does the lookup work on your behalf.
- **Record types** — `A` (name → IPv4), `AAAA` (→ IPv6), `CNAME` (alias → another name), `MX` (mail), `TXT` (misc/verification).
- **TTL (Time To Live)** — how long a DNS answer may be cached before re-checking.

> 🏭 **DE relevance:** A pipeline that suddenly throws `Could not resolve host` / `NameResolutionError` has a **DNS** problem — the API's domain didn't resolve (typo in URL, DNS outage, VPN/proxy misconfig, or the service moved). This is a *different* failure class from "connection refused" (TCP) or "401" (HTTP). Knowing the layer tells you where to look. DNS results are cached, so a stale cache can also send you to an old IP.


## 2.5 Step 2 — TCP: The Reliable Pipe (3-Way Handshake)

Now we have an IP. **TCP** (Transmission Control Protocol) builds a **reliable, ordered** connection between client and server. HTTP/1.x and /2 ride on top of TCP.

Why TCP? Networks lose and reorder packets. TCP guarantees your bytes arrive **completely and in order** (it re-sends lost packets, reassembles them). It's the difference between a reliable courier that confirms delivery vs. tossing postcards into the wind.

**The 3-way handshake** (how a TCP connection opens):
```
   CLIENT                                   SERVER
      │ ── SYN ──────────────────────────▶  │   "Hi, can we talk? (sequence x)"
      │ ◀───────────────── SYN-ACK ───────  │   "Sure, got it. (sequence y, ack x+1)"
      │ ── ACK ──────────────────────────▶  │   "Great, confirmed. (ack y+1)"
      │                                      │
      │ === connection established ===       │   now HTTP data can flow
```
- **SYN** = "synchronize" (let's start).
- **SYN-ACK** = "synchronize + acknowledge" (I agree).
- **ACK** = "acknowledge" (confirmed).

**Ports:** the IP finds the *machine*; the **port** finds the *program* on it. HTTP defaults to **port 80**, HTTPS to **port 443**. `api.github.com:443` = "the HTTPS service on that host".

> 🧠 **Interview tip:** *"TCP vs UDP?"* → TCP is reliable, ordered, connection-oriented (used by HTTP/1-2, databases, file transfer). UDP is fast, connectionless, no delivery guarantee (used by video/voice, DNS queries, and HTTP/3's QUIC).

> 🏭 **DE relevance:** `ConnectionRefusedError` / `Connection reset` means TCP couldn't establish or was dropped — the host resolved (DNS ok) but nothing is listening on that port, a firewall blocked it, or the service is down. Again: a *different* layer than a DNS error or an HTTP 500.


## 2.6 Step 3 — TLS: The Encryption Handshake (HTTPS)

If the URL is `https://`, after TCP connects we do a **TLS handshake** to create an encrypted tunnel. Everything after flows encrypted. (Recall Module 1.21: **TLS** is the modern protocol; **SSL** is its deprecated ancestor.)

TLS achieves three things:
1. **Encryption** — eavesdroppers on the network see gibberish.
2. **Authentication** — the server proves it's really `api.github.com` (via a **certificate** signed by a trusted **Certificate Authority / CA**).
3. **Integrity** — tampering with data in transit is detected.

**Simplified TLS 1.2 handshake:**
```
   CLIENT                                         SERVER
     │ ── ClientHello ──────────────────────────▶ │  "I support these cipher suites + TLS versions"
     │ ◀─ ServerHello + Certificate ─────────────  │  "Here's my chosen cipher + my CA-signed cert"
     │                                             │
     │  [client verifies certificate against       │
     │   trusted CAs; checks name + expiry]         │
     │                                             │
     │ ── key exchange (e.g. ECDHE) ─────────────▶ │  both derive the same secret session key
     │ ◀─ Finished ──────────────────────────────  │
     │ === encrypted tunnel ready ===              │  HTTP now flows encrypted
```

Key ideas explained simply:
- **Asymmetric crypto** (public/private key) is used *once* to safely agree on a shared secret. It's secure but slow.
- **Symmetric crypto** (one shared session key) then encrypts the actual data. It's fast.
- **Certificate** = a digital ID card for the server, signed by a CA your OS/browser already trusts. If the name doesn't match, it's expired, or the CA isn't trusted → the handshake fails.
- **TLS 1.3** (modern) streamlines this to fewer round-trips = faster.

> 🏭 **DE relevance — the errors you WILL see:**
> - `SSLCertVerificationError: certificate has expired` → the API's cert expired; wait for them to renew (don't disable verification in production!).
> - `CERTIFICATE_VERIFY_FAILED: unable to get local issuer certificate` → your machine is missing CA certs (common in fresh Docker images → install `ca-certificates`).
> - Tempting "fix" `verify=False` in `requests` **disables security** and exposes secrets/tokens — an org red flag. Fix the certs instead.


## 2.7 Step 4 — The HTTP Request (Anatomy, Line by Line)

The tunnel is open. Now the client sends an actual HTTP request. Here is a **real HTTP/1.1 request as raw text** (this is literally what travels over the wire):

```
POST /api/v1/employees HTTP/1.1          ← 1. REQUEST LINE  (method + path + version)
Host: api.company.com                    ← 2. HEADERS begin (metadata, key: value)
Authorization: Bearer eyJhbGciOiJ...
Content-Type: application/json
Accept: application/json
Content-Length: 44
User-Agent: python-requests/2.32.0
                                         ← 3. BLANK LINE  (separates headers from body)
{"name": "Asha", "department": "Data"}   ← 4. BODY / PAYLOAD  (only for POST/PUT/PATCH)
```

Four parts (this exact structure is asked in interviews):

1. **Request line** = `METHOD  PATH  HTTP-VERSION`.
   - `POST` — the method (intent). See 2.8.
   - `/api/v1/employees` — the path (which resource). `v1` is versioning (Module 1.17).
   - `HTTP/1.1` — the protocol version.
2. **Headers** — `Key: Value` lines carrying metadata *about* the message (who, what format, how long, auth).
3. **Blank line** — a single empty line marks "headers finished, body begins". Missing it breaks the message.
4. **Body / payload** — the actual data. GET/HEAD/DELETE usually have **no** body; POST/PUT/PATCH carry one (JSON here).

> 🧠 **Interview tip:** Be able to sketch these 4 parts from memory. A surprising number of candidates can't, and it's the fastest way to prove you truly understand HTTP.


## 2.8 HTTP Methods (Overview — full deep dive in Module 9)

The **method** (a.k.a. **verb**) states your *intent*. Here's the working set; Module 9 gives each its own example, diagram, and interview question.

| Method | Intent | Has body? | Safe? | Idempotent? |
|--------|--------|-----------|-------|-------------|
| **GET** | Read a resource | No | ✅ | ✅ |
| **POST** | Create a resource / submit data | Yes | ❌ | ❌ |
| **PUT** | Replace a resource entirely | Yes | ❌ | ✅ |
| **PATCH** | Partially update a resource | Yes | ❌ | ⚠️ usually no |
| **DELETE** | Remove a resource | No/opt | ❌ | ✅ |
| **HEAD** | Like GET but headers only (no body) | No | ✅ | ✅ |
| **OPTIONS** | Ask what methods/headers are allowed (CORS preflight) | No | ✅ | ✅ |
| **TRACE** | Echo the request (debug; usually disabled for security) | No | ✅ | ✅ |
| **CONNECT** | Establish a tunnel (used by proxies for HTTPS) | No | ❌ | ❌ |

Recall the definitions (Module 1.16):
- **Safe** = read-only, changes nothing on the server (GET, HEAD, OPTIONS).
- **Idempotent** = calling it N times has the same effect as calling it once (GET, PUT, DELETE, HEAD, OPTIONS).

> 🧠 **Interview trap:** *"Is DELETE idempotent even though the second call returns 404?"* → **Yes.** Idempotency is about the **server's final state** (the resource is gone either way), not about identical response codes.

> 🏭 **DE relevance:** Extraction is almost all **GET** (with pagination/filter query params). Loading into an API sink uses **POST** (create events) or **PUT/PATCH** (upsert-style updates). Choosing PUT/upsert over POST where possible makes retries idempotent (Module 1.16, Module 17).


## 2.9 HTTP Headers (the ones you MUST know)

Headers are `Key: Value` metadata. There are hundreds; here are the ones a Data Engineer actually uses. (Header names are **case-insensitive**.)

### Common REQUEST headers (client → server)
| Header | Meaning | DE example |
|--------|---------|------------|
| `Host` | Which domain you want (one IP can host many sites) | Required in HTTP/1.1 |
| `Authorization` | Credentials — `Bearer <token>` or `Basic <base64>` | Sending your API token |
| `Content-Type` | Format of the **body you're sending** | `application/json` on POST |
| `Accept` | Format you **want back** (content negotiation, 1.18) | `Accept: application/json` |
| `Accept-Encoding` | Compression you support | `gzip` for big pulls (1.18) |
| `User-Agent` | Who's calling (client identity) | `python-requests/2.32.0` |
| `Content-Length` | Size of the body in bytes | Set automatically |
| `Cookie` | Cookies the browser sends back | (browser-driven, 2.12) |
| `If-None-Match` / `If-Modified-Since` | Conditional fetch for caching | Skip re-download if unchanged |

### Common RESPONSE headers (server → client)
| Header | Meaning | DE example |
|--------|---------|------------|
| `Content-Type` | Format of the **body being returned** | `application/json; charset=utf-8` |
| `Content-Length` | Body size | — |
| `Content-Encoding` | Compression applied | `gzip` (auto-decompressed by `requests`) |
| `Set-Cookie` | Server asks client to store a cookie | Session id (2.12) |
| `Cache-Control` / `ETag` | Caching directives (1.17) | `ETag: "abc"` for conditional GETs |
| `Location` | Where a created/redirected resource lives | `201 Created` → URL of new row |
| `Retry-After` | How long to wait after 429/503 | **Read this to back off!** (Module 17) |
| `X-RateLimit-Limit` / `-Remaining` / `-Reset` | Rate-limit budget (1.17) | Throttle before hitting 429 |
| `WWW-Authenticate` | How to authenticate after a 401 | Tells you the scheme (Bearer) |

```
   Custom / non-standard headers conventionally start with X-  (e.g. X-Request-Id)
   → great for tracing a single request across microservices/logs.
```

> 🏭 **DE relevance:** Your extractor should *read* `X-RateLimit-Remaining` and `Retry-After` to pace itself, and *set* `Authorization`, `Accept: application/json`, and `Accept-Encoding: gzip`. Logging the `X-Request-Id` from responses lets you correlate a failed pull with the provider's support team.


## 2.10 Step 5 — The HTTP Response (Anatomy, Line by Line)

The server does its work and sends back a response. Raw HTTP/1.1 response:

```
HTTP/1.1 201 Created                     ← 1. STATUS LINE  (version + code + reason)
Content-Type: application/json           ← 2. HEADERS
Location: /api/v1/employees/42
Content-Length: 58
X-RateLimit-Remaining: 59
                                         ← 3. BLANK LINE
{"id": 42, "name": "Asha", "department": "Data"}   ← 4. BODY / PAYLOAD
```

Four parts, mirroring the request:
1. **Status line** = `HTTP-VERSION  STATUS-CODE  REASON-PHRASE`. `201 Created` = success, a resource was created. The **reason phrase** ("Created") is human-friendly text; code drives logic.
2. **Headers** — metadata about the response (format, size, caching, rate-limit, `Location` of the new resource).
3. **Blank line** — separates headers from body.
4. **Body / payload** — the returned data (JSON here). Some responses have no body (e.g., `204 No Content`).

> 🧠 **Interview tip:** *"After a successful POST that creates a resource, what status and header do you expect?"* → **201 Created** with a **`Location`** header pointing to the new resource's URL. Small detail, big signal.


## 2.11 Status Codes — The Complete Catalog

Recall the five families (Module 1.13). Here's the working catalog with the ones you'll actually meet. Memorize the **bold** ones cold.

### 1xx — Informational ("hold on")
| Code | Name | Meaning |
|------|------|---------|
| 100 | Continue | "Send the rest of the body." |
| 101 | Switching Protocols | e.g., upgrading to WebSocket. |

### 2xx — Success ("here you go")
| Code | Name | When |
|------|------|------|
| **200** | OK | Standard success (GET/PUT/PATCH). |
| **201** | Created | Resource created (POST) — expect a `Location` header. |
| 202 | Accepted | Request accepted, processing **asynchronously** (great for long ETL jobs). |
| **204** | No Content | Success, **no body** (common for DELETE). |
| 206 | Partial Content | Range request (resumable/partial downloads). |

### 3xx — Redirection ("look elsewhere")
| Code | Name | When |
|------|------|------|
| 301 | Moved Permanently | Resource has a new permanent URL. |
| 302 | Found | Temporary redirect. |
| **304** | Not Modified | Your cached copy is still fresh (conditional GET) — saves bandwidth. |
| 307/308 | Temporary/Permanent Redirect | Redirect **without** changing the method. |

### 4xx — Client errors ("YOU messed up") — fix your request
| Code | Name | When (and what to do) |
|------|------|------------------------|
| **400** | Bad Request | Malformed request/JSON, bad params → fix the request. |
| **401** | Unauthorized | Missing/invalid credentials (AuthN) → check token. |
| **403** | Forbidden | Authenticated but not allowed (AuthZ) → check permissions/scopes. |
| **404** | Not Found | Resource/endpoint doesn't exist → check URL/id. |
| 405 | Method Not Allowed | Wrong verb for this endpoint (e.g., POST to a read-only URL). |
| 406 | Not Acceptable | Server can't produce the `Accept` format. |
| 409 | Conflict | State conflict (duplicate key, version clash). |
| 410 | Gone | Resource permanently removed. |
| 415 | Unsupported Media Type | Bad `Content-Type` (e.g., sent XML, server wants JSON). |
| 422 | Unprocessable Entity | Syntactically ok but **validation** failed (very common in APIs). |
| **429** | Too Many Requests | **Rate limited** → read `Retry-After`, back off (Module 17). |

### 5xx — Server errors ("THEY messed up") — often retryable
| Code | Name | When (and what to do) |
|------|------|------------------------|
| **500** | Internal Server Error | Server crashed handling a valid request → retry w/ backoff, then report. |
| 501 | Not Implemented | Server doesn't support the method. |
| **502** | Bad Gateway | A proxy/gateway got a bad response from upstream (recall 1.20). |
| **503** | Service Unavailable | Overloaded/down for maintenance → honor `Retry-After`. |
| 504 | Gateway Timeout | Upstream too slow behind a gateway. |

> 🧠 **Interview gold (repeat until automatic):** 2xx = success, 3xx = redirect, **4xx = client's fault, 5xx = server's fault.** 401 vs 403, 400 vs 422 (bad syntax vs failed validation), 502 vs 503 (bad upstream vs unavailable).

> 🏭 **DE retry policy (this is the whole game in Module 17):**
> ```
>   2xx           → success, parse the data
>   429           → sleep for Retry-After (or backoff), then retry
>   500/502/503/504 → transient → retry with exponential backoff + jitter
>   400/401/403/404/422 → your bug → DO NOT blindly retry; log, fix, or alert
> ```


## 2.12 Cookies (How They Really Work)

HTTP is stateless, so how does a site remember you're logged in across requests? **Cookies** (recall Module 1.14 — here's the mechanism).

```
   1. You log in:
      CLIENT ── POST /login (user+pass) ──▶ SERVER
      CLIENT ◀── 200 OK
                 Set-Cookie: session_id=xyz789; HttpOnly; Secure; SameSite=Lax; Max-Age=3600
                 ──────────── SERVER
              (browser stores this cookie)

   2. Every later request, the browser AUTOMATICALLY attaches it:
      CLIENT ── GET /dashboard
                Cookie: session_id=xyz789 ──▶ SERVER
              (server recognizes you)
```

Important cookie attributes (security interview material):
| Attribute | Effect |
|-----------|--------|
| `HttpOnly` | JavaScript **cannot** read it → mitigates XSS token theft. |
| `Secure` | Only sent over **HTTPS**. |
| `SameSite=Lax/Strict/None` | Controls sending on cross-site requests → mitigates CSRF. |
| `Max-Age` / `Expires` | Lifetime; without them it's a **session cookie** (dies when browser closes). |
| `Domain` / `Path` | Which URLs the cookie is sent to. |

> ⚠️ **Key distinction for DE:** Cookies are **browser-centric**. Server-to-server pipelines (`requests`, Airflow) almost always use **tokens in the `Authorization` header**, not cookies. If an API *requires* cookies, you handle them with a `requests.Session()` (which stores cookies for you) — but that's the exception, not the norm.


## 2.13 Sessions (Server-Side State)

A **session** is memory the *server* keeps about a client, tied to a cookie that holds only a small **session id**.

```
   Browser cookie:  session_id = xyz789          (tiny, opaque id)
                          │
                          ▼
   Server store (memory / Redis / DB):
     xyz789 → { user_id: 42, name: "Asha", cart: [ ... ], logged_in_at: ... }
```

Flow: on login the server creates a session record + returns its id as a cookie; on each request it looks up the id to recover *who you are*.

**Session vs Token (the interview matchup):**
| | Session (server-side) | Token / JWT (client-side) |
|---|------------------------|----------------------------|
| Where state lives | On the server | Inside the token itself |
| Scales horizontally? | Harder (need shared store like Redis) | Easy (**stateless** — any server can verify) |
| Revocation | Easy (delete the session) | Harder (token valid until expiry unless you keep a denylist) |
| Typical use | Traditional server-rendered websites | **APIs, mobile, microservices** |

> 🏭 **DE relevance:** Distributed systems favor **stateless tokens** precisely because you can load-balance freely (recall Module 1.16 & 1.20). That's why almost every data API you consume uses tokens, not sessions.


## 2.14 Tokens & JWT (Structure, Signing, Verification)

A **token** is a string that proves who you are / what you may do, sent in `Authorization: Bearer <token>`. The most common format is **JWT** (**JSON Web Token**, pronounced "jot").

A JWT is **three Base64URL-encoded parts joined by dots**: `header.payload.signature`
```
   eyJhbGciOiJIUzI1NiIsInR5cCI6IkpXVCJ9  .  eyJzdWIiOiI0MiIsIm5hbWUiOiJBc2hhIn0  .  4pE7...signature
   └──────────── HEADER ───────────────┘    └──────────── PAYLOAD ─────────────┘    └──── SIGNATURE ────┘
```

Decoded:
```
   HEADER   { "alg": "HS256", "typ": "JWT" }          ← which signing algorithm
   PAYLOAD  { "sub": "42", "name": "Asha",            ← "claims": data about the user
              "role": "data_engineer",
              "iat": 1735689600,                       ← issued-at (unix time)
              "exp": 1735693200 }                      ← EXPIRY (unix time) ← critical!
   SIGNATURE  HMAC-SHA256( base64(header) + "." + base64(payload), SECRET_KEY )
```

**How verification works (why JWT is tamper-proof but not secret):**
1. Server creates the token and **signs** header+payload with a secret key → the signature.
2. Client sends the token back on each request.
3. Server **recomputes** the signature from the received header+payload using its secret. If it matches, the token is authentic and untampered.

> ⚠️ **Two things everyone gets wrong:**
> 1. **JWT payload is encoded, NOT encrypted.** Anyone can Base64-decode and read the claims. **Never put secrets (passwords, PII) in a JWT.** The signature prevents *tampering*, not *reading*.
> 2. **Tokens expire (`exp`).** A pipeline running at 3 AM will get **401** the moment the token expires mid-run unless you **refresh** it (2.15, Module 17).

> 🏭 **DE relevance:** You'll receive JWTs from OAuth flows, cache them, watch `exp`, and refresh before expiry. A robust extractor checks token validity *before each batch* and re-authenticates on 401 exactly once (then fails loudly if it still can't).


## 2.15 OAuth 2.0 & Bearer Tokens (the Flow)

**OAuth 2.0** is the industry standard for **delegated authorization** — letting an app act on your behalf **without** handing over your password.

**Analogy:** a **hotel key card**. Reception (the authorization server) checks your ID once, then gives you a *card* (token) that opens *only your room* (scoped access) and *expires at checkout*. You never give the cleaning staff your passport; you just show the card.

**"Bearer token"** literally means *"whoever bears (holds) this token gets access."* → so protect it like a password; anyone who steals it can use it until it expires.

### Client Credentials flow (the one Data Engineers use most — machine-to-machine)
No human/browser involved; a service authenticates with a client id + secret.
```
   ┌───────────────┐   1. POST /oauth/token                ┌────────────────────┐
   │ Your pipeline │   grant_type=client_credentials       │  Authorization     │
   │  (client)     │   client_id + client_secret + scope   │  Server (issuer)   │
   │               │ ─────────────────────────────────────▶│                    │
   │               │ ◀───────────────────────────────────  │                    │
   │               │   2. { access_token, expires_in,       └────────────────────┘
   │               │        token_type: "Bearer" }
   │               │
   │               │   3. call the API with the token:                 ┌──────────┐
   │               │   GET /data                                        │ Resource │
   │               │   Authorization: Bearer <access_token> ───────────▶│  Server  │
   │               │ ◀── 200 OK + data ─────────────────────────────────│  (API)   │
   └───────────────┘                                                    └──────────┘
       4. token nears expiry → repeat step 1 to REFRESH.
```

Core OAuth vocabulary:
| Term | Meaning |
|------|---------|
| **Resource Owner** | The user who owns the data (in client-credentials, it's the app itself). |
| **Client** | The app requesting access (your pipeline). |
| **Authorization Server** | Issues tokens (e.g., `login.microsoftonline.com`, Auth0). |
| **Resource Server** | The API holding the data. |
| **Access Token** | Short-lived token used on API calls. |
| **Refresh Token** | Longer-lived token used to get new access tokens without re-login. |
| **Scope** | The specific permissions granted (`read:orders`). |

> 🧠 **Interview tip:** *"OAuth vs JWT?"* → They're different layers. **OAuth 2.0 is a framework/flow** for *getting* a token; **JWT is a token format**. OAuth access tokens are *often* (not always) JWTs. Don't say "OAuth vs JWT" as if they compete.

> 🏭 **DE relevance:** Google, Microsoft, Salesforce, and most enterprise APIs use OAuth **client-credentials** for service accounts. Your pipeline: (1) exchange client_id/secret for an access token, (2) use it until near `expires_in`, (3) refresh. **Token refresh failures are one of the top causes of overnight pipeline breakage** — Module 17 codes this defensively.

Let's make the abstract concrete. We'll inspect the request/response anatomy live with `requests`.

````python
"""Module 2 demo: inspect the full HTTP request/response anatomy."""
import requests  # HTTP client (does DNS+TCP+TLS+HTTP for you under the hood)

# We send explicit headers so we can SEE content negotiation & identity (2.9).
headers = {
    "Accept": "application/json",        # we WANT JSON back (content negotiation, 1.18/2.9)
    "Accept-Encoding": "gzip",           # we accept gzip compression (1.18) for big pulls
    "User-Agent": "flask-masterclass/1.0 (learning)",  # identify ourselves (2.9)
}

# GET = the "read" method (2.8). timeout ALWAYS set (Module 1.19) so we never hang.
resp = requests.get("https://httpbin.org/get", headers=headers, timeout=10)

# ---- THE RESPONSE ANATOMY (2.10) ----
print("STATUS LINE   :", resp.status_code, resp.reason)   # e.g. 200 OK  (2xx = success)
print("HTTP VERSION  :", resp.raw.version)                 # 11 = HTTP/1.1, 20 = HTTP/2

print("\n--- RESPONSE HEADERS (metadata about the body) ---")
for k, v in resp.headers.items():
    print(f"  {k}: {v}")

print("\n--- BODY (payload), deserialized from JSON to a dict (Module 4) ---")
data = resp.json()
# httpbin echoes back the request it saw — so we can inspect what WE sent:
print("  Server saw our headers:")
for k, v in data["headers"].items():
    print(f"    {k}: {v}")
````

- `resp.headers.items()` — the **response headers** (2.9/2.10): note `Content-Type`, maybe `Content-Encoding: gzip` (auto-decompressed for you).
- `resp.json()` — **deserialize** the JSON **body/payload** into a Python `dict` (Module 4).
- `data["headers"]` — `httpbin.org` echoes back the request it received, so you can literally *see the request you sent* — a fantastic learning/debugging trick.

## ✅ Module 2 — Summary (revise before interviews)

- **HTTP** = HyperText Transfer Protocol: a **request/response**, **stateless**, (in 1.x) **text-based** protocol. **HTTPS** = HTTP inside a **TLS** encrypted tunnel.
- **Versions:** 1.0 (connection per request) → 1.1 (keep-alive) → 2 (multiplexing, binary, header compression) → 3 (QUIC over UDP). Multiplexing kills head-of-line blocking.
- **The journey:** URL parse → **DNS** (name→IP) → **TCP** (3-way handshake: SYN/SYN-ACK/ACK) → **TLS** (verify cert, agree keys) → **HTTP request** → server work → **HTTP response** → use/render.
- **Ports:** HTTP=80, HTTPS=443. **IP finds the machine, port finds the program.**
- **Request anatomy:** request line (method + path + version) → headers → blank line → body. **Response anatomy:** status line (version + code + reason) → headers → blank line → body.
- **Methods:** GET(read/safe/idempotent), POST(create), PUT(replace/idempotent), PATCH(partial), DELETE(idempotent), HEAD(headers only), OPTIONS(capabilities/CORS), TRACE, CONNECT.
- **Headers you must know:** `Authorization`, `Content-Type` (what I'm sending), `Accept` (what I want), `Accept-Encoding`, `User-Agent`, `Set-Cookie`/`Cookie`, `Cache-Control`/`ETag`, `Location`, `Retry-After`, `X-RateLimit-*`, `WWW-Authenticate`.
- **Status codes:** 1xx info, 2xx success (200/201/202/204), 3xx redirect (301/304), 4xx **client** error (400/401/403/404/409/422/429), 5xx **server** error (500/502/503/504). 401=AuthN, 403=AuthZ; 400=bad syntax, 422=failed validation.
- **Cookies:** server `Set-Cookie` → browser stores → auto-sends `Cookie`. Attributes: `HttpOnly`, `Secure`, `SameSite`, `Max-Age`. **Sessions:** server-side state keyed by a session-id cookie.
- **JWT:** `header.payload.signature`, Base64URL-encoded (**readable, not secret**), signature proves it's untampered. Watch `exp` (expiry). Never store secrets in the payload.
- **OAuth 2.0:** framework for delegated authorization; **client-credentials** flow is the DE default (client_id+secret → access_token → call API → refresh). **Bearer token** = "holder gets access". OAuth = flow, JWT = format (often combined).
- **Master model:** every failure maps to a **layer** — DNS / TCP / TLS / HTTP(status family). Diagnose the layer first.


## 🎤 Module 2 — Interview Questions (with model answers)

1. **Walk me through what happens when you type a URL and press Enter.** URL parse → DNS resolves host to IP → TCP 3-way handshake → (if https) TLS handshake verifies cert & agrees keys → send HTTP request → server routes/authenticates/queries → HTTP response → client renders/parses.
2. **HTTP vs HTTPS?** HTTPS is HTTP carried inside a TLS-encrypted tunnel; it adds confidentiality, server authentication (via CA-signed certificate), and integrity.
3. **What does DNS do and what error do you see when it fails?** Translates a domain name to an IP address; failure shows as "could not resolve host"/`NameResolutionError`.
4. **Explain the TCP 3-way handshake.** SYN (client asks) → SYN-ACK (server agrees) → ACK (client confirms); establishes a reliable, ordered connection.
5. **TCP vs UDP?** TCP is reliable, ordered, connection-oriented (HTTP/1-2, DBs); UDP is fast, connectionless, no guarantees (streaming, DNS, HTTP/3's QUIC).
6. **What are the three guarantees TLS provides?** Encryption (privacy), authentication (server identity via certificate), and integrity (tamper detection).
7. **Why is `verify=False` dangerous?** It disables TLS certificate verification, enabling man-in-the-middle attacks that can steal your tokens/data; fix the CA/cert instead.
8. **Name the four parts of an HTTP request and response.** Request: request line, headers, blank line, body. Response: status line, headers, blank line, body.
9. **What improved in HTTP/2?** Multiplexing (parallel requests on one connection), binary framing, header compression (HPACK) — eliminating HTTP-level head-of-line blocking.
10. **`Content-Type` vs `Accept`?** `Content-Type` describes the body of the *current* message; `Accept` tells the server the format the client *wants back*.
11. **401 vs 403 vs 404?** 401 = not authenticated; 403 = authenticated but not permitted; 404 = resource/endpoint not found.
12. **400 vs 422?** 400 = malformed/bad request (can't even parse it); 422 = well-formed but fails business/validation rules.
13. **What does 429 mean and how do you handle it in a pipeline?** Too Many Requests (rate limited); read `Retry-After`, throttle, and use exponential backoff with jitter.
14. **Which 5xx codes are typically retryable?** 500/502/503/504 (transient server/gateway issues) — retry with backoff; do not retry most 4xx.
15. **How do cookies keep a user logged in?** Server sends `Set-Cookie`; the browser stores it and auto-sends `Cookie` on subsequent requests so the server recognizes the session.
16. **Session vs token authentication — which scales better and why?** Tokens (JWT): they're stateless (self-contained), so any server behind a load balancer can verify without a shared session store.
17. **What are the three parts of a JWT and is the payload secret?** header.payload.signature; the payload is Base64URL-**encoded**, not encrypted — readable by anyone — so never put secrets in it.
18. **How does a server verify a JWT hasn't been tampered with?** It recomputes the signature over the received header+payload using its secret key and compares it to the token's signature.
19. **OAuth vs JWT — do they compete?** No: OAuth 2.0 is a framework/flow for obtaining authorization tokens; JWT is a token *format*. OAuth access tokens are often JWTs.
20. **Describe the OAuth client-credentials flow for a data pipeline.** POST client_id+client_secret (+scope) to the token endpoint → receive a short-lived Bearer access_token → call the API with `Authorization: Bearer <token>` → refresh before `expires_in` elapses.
21. **Is DELETE idempotent if the second call returns 404?** Yes — idempotency concerns the server's final state (resource gone), not identical response codes.
22. **What's the significance of the blank line in an HTTP message?** It separates headers from the body; without it, parsers can't tell where headers end.

## ⚠️ Module 2 — Common Mistakes (and how to avoid them)

1. **Thinking JWT payloads are secret.** They're only Base64-encoded → fully readable. Never store passwords/PII in a JWT; the signature protects *integrity*, not *confidentiality*.
2. **Ignoring token expiry.** `exp` passes mid-run → sudden 401s at 3 AM. Track expiry and refresh proactively (Module 17).
3. **Disabling TLS verification (`verify=False`) to "make it work".** This removes security and can leak tokens via MITM. Fix the certificate/CA bundle instead.
4. **Confusing 401 and 403 (again).** 401 = who are you? (bad/missing creds). 403 = I know you, not allowed. Different fixes: credentials vs permissions/scopes.
5. **Confusing 400 and 422.** 400 = can't parse your request at all; 422 = parsed fine but failed validation rules. They point to different bugs.
6. **Retrying non-retryable errors.** Retrying a 400/401/404 just repeats a broken request. Retry only 429/5xx/timeouts, with backoff.
7. **Not honoring `Retry-After`.** On 429/503 the server tells you how long to wait — ignoring it can get you rate-limited harder or IP-banned.
8. **Forgetting the `Host` header / assuming one IP = one site.** Many domains share an IP; `Host` (and TLS SNI) select the right site. Rarely an issue with `requests` (set for you), but explains virtual hosting in interviews.
9. **Believing HTTP is stateful.** It isn't — cookies/sessions/tokens *simulate* memory on top of a stateless protocol.
10. **Treating "SSL" as separate/current.** It's the deprecated predecessor of TLS; you're using TLS. Say TLS in interviews.
